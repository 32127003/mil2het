```python
import scanpy as sc
covid_path ="./covid/covid_data.h5ad"
```

```python
covid = sc.read_h5ad(covid_path)
print(covid)
```

```python
#  remove genes with nonzero expressions in fewer than five cells, normalize the total gene expression counts of each cell to sum up to 1e4, and log-transform the data using the natural logarithm.



sc.pp.filter_genes(covid, min_cells=5)
for_removal = ['COVID19_Participant34','COVID_Participant33','COVID19_Participant21','Control_Participant20','COVID19_Participant19','COVID19_Participant18','Control_Participant17','COVID19_Participant16',]

# donor id가 for_removal에 있으면 제거한다.
covid = covid[~covid.obs['donor_id'].isin(for_removal)]


```

```python
# SARSCoV2_PCR_Status 가 pos면 1 neg면 0으로 mapping해서 'label' column 생성 후 저장
covid.obs['label'] = covid.obs['SARSCoV2_PCR_Status'].map({'pos': 1, 'neg': 0})
covid.write_h5ad("./covid/covid_data.h5ad")
```

```python
# log normalize 된 것인지 확인. yes no로 출력
if covid.X.max() > 100:
    print("No, the data is not log normalized.")
else:
    print("Yes, the data is log normalized.")
```
