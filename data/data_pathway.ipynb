```python
reactome_path='./pathway/c2.cp.reactome.v2026.1.Hs.symbols.gmt'
hallmark_path='./pathway/h.all.v2026.1.Hs.symbols.gmt'
```

```python
# preprocessing. 형식: {pathway_name: [gene1, gene2, ...]} dict여야 함. pickle 파일로 저장.
import pickle

def load_gmt_to_dict(gmt_path, min_size=10, max_size=300):
    pathway_dict = {}

    with open(gmt_path, "r") as f:
        for line in f:
            parts = line.strip().split("\t")
            if len(parts) < 3:
                continue

            pathway_name = parts[0]
            genes = parts[2:]  # skip description

            # optional size filtering
            if min_size <= len(genes) <= max_size:
                pathway_dict[pathway_name] = genes

    return pathway_dict


reactome = load_gmt_to_dict(reactome_path)
hallmark = load_gmt_to_dict(hallmark_path)

with open("reactome_pathway.pkl", "wb") as f:
    pickle.dump(reactome, f)

with open("hallmark_pathway.pkl", "wb") as f:
    pickle.dump(hallmark, f)

print(len(reactome))
print(len(hallmark))
print(list(reactome.items())[:3])
print(list(hallmark.items())[:3])
```

```python
# common_genes.tsv에 없으면 pathway에서 제거하기
with open("common_genes.tsv", "r") as f:
    common_genes = set(line.strip() for line in f)
def filter_pathway_dict(pathway_dict, valid_genes):
    filtered_dict = {}
    for pathway, genes in pathway_dict.items():
        filtered_genes = [gene for gene in genes if gene in valid_genes]
        if filtered_genes:
            filtered_dict[pathway] = filtered_genes
    return filtered_dict
reactome_filtered = filter_pathway_dict(reactome, common_genes)
hallmark_filtered = filter_pathway_dict(hallmark, common_genes)
with open("reactome_pathway_filtered.pkl", "wb") as f:
    pickle.dump(reactome_filtered, f)
with open("hallmark_pathway_filtered.pkl", "wb") as f:
    pickle.dump(hallmark_filtered, f)
print(len(reactome_filtered))
print(len(hallmark_filtered))
```
